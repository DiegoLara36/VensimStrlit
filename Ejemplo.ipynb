```python
import pysd
import matplotlib.pyplot as plt
import pandas as pd
import os
```

```python
modelo= pysd.read_vensim('Modelos/Prob01-Publicidad.mdl')


valores = modelo.run(return_columns=['Usuarios Activos'])
tabla = valores.head(15)  ## 10 filas
print(" Caso: Modelo de Adopción, Desgaste y Recuperación de Clientes")
print("-" * 60)
# IMprime la tabla


valores.plot()
plt.text(1, 1, ' prob01 ', fontsize=12, color='red')
plt.title("Modelo adopción y desgaste publicidad", fontsize=22, color='green')
plt.ylabel('Usuario Potenciales')
plt.xlabel('Meses')
##plt.legend(loc
plt.show()
```

```python
modelo= pysd.read_vensim('Modelos/Prob02-Inventario.mdl')


valores = modelo.run(return_columns=['Inventario en Almacen'])
tabla = valores.head(15)  ## 10 filas
print(" Caso: Modelo de Gestión de Inventario con Retraso de Proveedor")
print("-" * 60)
# IMprime la tabla


valores.plot()

plt.text(1, 1, ' prob 02 ', fontsize=12, color='red')

plt.title("Modelo de Gestión de Inventario con Retraso",
          fontsize=22, color='yellow')

plt.ylabel('Unidades')
plt.xlabel('Dias')

plt.grid(True, linestyle='--', alpha=0.7)

plt.xlim(left=0)
plt.ylim(bottom=0)

plt.show()
```
